* __Task 1__: Calculate the product of nonzero elements on the diagonal of a rectangular matrix.  
 For example, for X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) answer is 3.

```python
# numpy implementation 

X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])
# np.diag help us to find diagonal
diagonal=np.diag(X)
# we jsut multiply all nums where nonzero
res = np.prod(diagonal[diagonal!=0])
print(f'np_res {res}')

# python implementation 
X_list = X.tolist()
py_res=1

# we need to make sure our matrix is square if not only taking square part
for i in range(min(len(X_list),len(X_list[0]))):
    if X_list[i][i]!=0:
        py_res*=X_list[i][i]
print(f'python {py_res}')
```

* __Task 3__: Find the maximum element in the vector x among the elements preceded by zero. 
 For example, for x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) answer is 5.

```python
# python implementation 
x = [6, 2, 0, 3, 0, 0, 5, 7, 0]

# we will loop through and get value behind 0 and add it to newlist and we get max at the end
res_py=max(x[i] for i in range(1,len(x)) if x[i-1]==0)

print(f'python: {res_py}')

# numpy implementation
# x[:-1] returns us values except last element
# x[1:] returns values except first , try to print them and u will understand why this works
x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])

print(f'numpy: {x[1:][x[:-1]==0].max()}')
```

* __Task 4__: Implement run-length encoding. For some vector x, it is necessary to return a tuple of two vectors of the same length. The first one contains numbers and the second one contains how many times they need to be repeated.  
 For example, for x = np.array([2, 2, 2, 3, 3, 3, 5]) answer is (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
x = [2, 2, 2, 3, 3, 3, 5]
values = []
counts = []
c=1
for idx in range(0,len(x)):
    if len(x)-1>idx and x[idx]==x[idx+1]:
        c+=1
    else:
        values.append(x[idx])
        counts.append(c)
        c=1
print(f'python : {list(values),counts}')


#numpy impelemtatiuon 


# we first need to find a and b whcih a consist all array without first element and b without last element
x = [2, 2, 2, 3, 3, 3, 5]
x = np.array(x)
a,b = x[1:],x[:-1] # for thsi to work x must be numpy array

# now we compare these a and b and get true or false array, so true shows us where is changin vaue started 

v=np.r_[True,a!=b,True] # we need true in start and end cause true shows us where repeating starts and ends
#np.r_ just add them correctly 
# i just store to var v 
# now we find COUNT , first we gotta find indexes of true vals
v = np.flatnonzero(v) # flatnonzer returns idx of nonzero val
counts_np=np.diff(v) # diff returns value by distracting nums next to each other

# now we find values, for this we just return value idx in v
values_np=x[v[:-1]]
print(f'numpy: {values_np,counts_np}')
```

* __Task 5__: Two samples of objects are given - X and Y. Calculate the matrix of Euclidean distances between objects. Compare with function scipy.spatial.distance.cdist by operating speed.

```python
import timeit

X = np.array([[0, 0],
              [1, 1]])       

Y = np.array([[3, 4],
              [0, 1],
              [1, 0]])        

# we use broadcasting here 
# we add None ( which is one dimension to X) and we can subtract Y from it easily

t_np=timeit.timeit(lambda:np.sqrt(
    ((X[:,None]-Y) ** 2).sum(axis=2)
),number=1000)

#__________________________________

from scipy.spatial.distance import cdist

t_cp=timeit.timeit(lambda:cdist(X,Y,metric="euclidean"),number=1000)




#------------------------------
import math

def dist_py(X, Y):
    return [[math.sqrt(sum((a - b) ** 2 for a, b in zip(p, q))) for q in Y] for p in X]


t_py = timeit.timeit(lambda: dist_py(X.tolist(), Y.tolist()), number=1000)



print(f'python {t_py} seconds')
print(f'numpy {t_np} seconds')
print(f'scipy {t_cp} seconds')


# runned several times and scipy is being more faster 
```

2. One serving of CrunchieMunchies contains 60 calories. How much higher is the average calorie count of your competitors?

Store the answer in the variable average_calories and print the variable in the terminal

```python
# substract and get module of it and round it to look more familiar
average_calories=np.round(np.abs(60-np.average(calorie_stats)),1)
print(average_calories)
```

3. Does the average calorie count accurately reflect the distribution of the data set? Let's sort the data and see.

Sort the data and store the result in the calorie_stats_sorted variable. Print the sorted information.

Calculate the median of the data set and save your answer in median_calories. Print the median so you can see how it compares to the mean.

```python
median_calories = np.median(calorie_stats)
median_calories
```

5. While the median shows that at least half of our values ​​are over 100 calories, it would be more impressive to show that a significant portion of the competitors have higher calorie counts than CrunchieMunchies.

Calculate various percentiles and print them until you find the lowest percentile that exceeds 60 calories. Store this value in the nth_percentile variable.
